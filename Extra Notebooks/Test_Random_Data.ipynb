#### <span style="color:red">**Predict Hazard from Asteroid JSON Values**</span>

```python
import pandas as pd
import joblib
import requests
from pathlib import Path
```

<span style="color:blue">**Paste in your copied JSON below**</span>

```python
# replace the JSON below with your coied JSON data after the '=' sign
json_data = {
    "Absolute_Magnitude": 21.97518,
    "Est_Dia_In_Km_Min": 2.29749,
    "Est_Dia_In_Km_Max": 8.1854,
    "Relative_Velocity_Km_Per_Hr": 112282.21166,
    "Miss_Dist_Kilometers": 100715645.75149,
    "Minimum_Orbit_Intersection": 0.30902,
    "Jupiter_Tisserand_Invariant": 4.66746,
    "Epoch_Osculation": 2454095.02773,
    "Eccentricity": 0.54911,
    "Semi_Major_Axis": 3.28366,
    "Inclination": 5.32419,
    "Asc_Node_Longitude": 114.03363,
    "Orbital_Period": 449.79259,
    "Perihelion_Distance": 1.09642,
    "Perihelion_Arg": 83.48741,
    "Aphelion_Dist": 4.62313,
    "Perihelion_Time": 2458859.23533,
    "Mean_Anomaly": 189.13204,
    "Mean_Motion": 1.02875,
    "Hazardous": 0
}
```

```python
if json_data['Hazardous'] == 1:
    print("The asteroid is hazardous")
else:
    print("The asteroid is not hazardous")
```

<span style="color:red">**Prepare dataframe for prediction**</span>

```python
df = pd.DataFrame([json_data]).drop(columns=['Hazardous'])
df
```

<span style="color:red">**Load ML models**</span>

```python
CWD = Path.cwd().parent
MODEL = CWD / 'ML Models'
HAZARD_MODEL_PATH = MODEL / 'hazard_model.pkl'
SCALER_PATH = MODEL / 'scaler.pkl'


hazard_model = joblib.load(HAZARD_MODEL_PATH)
scaler = joblib.load(SCALER_PATH)
```

<span style="color:red">**Transform data to fit model**</span>

```python
df_scaled = scaler.transform(df)
```

<span style="color:red">**Prediction**</span>

```python
prediction = hazard_model.predict(df_scaled)[0]
print("Prediction:", "Hazardous" if prediction == 1 else "Not Hazardous")
```

<span style="color:red">**Prediction confidence**</span>

```python
proba = hazard_model.predict_proba(df_scaled)[0][1]

print("Probability of being Safe:", 100-(round(proba * 100, 2)), "%")
print("Probability of being hazardous:", round(proba * 100, 2), "%")
```
